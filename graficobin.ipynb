```python
import numpy as np
import matplotlib.pyplot as plt
from scipy.stats import binom

# Parámetros
n = 5
p_values = [0.20, 0.80]
color_unico = 'steelblue'
x = np.arange(0, n + 1)

# Crear la figura con 2 subplots
fig, axes = plt.subplots(1, 2, figsize=(12, 6), sharey=True)

for i, p in enumerate(p_values):
    # Cálculo de la PMF
    dist = binom.pmf(x, n, p)
    
    # Graficar
    bars = axes[i].bar(x, dist, color=color_unico, edgecolor='black', alpha=0.7)
    
    # Agregar las etiquetas de probabilidad sobre cada barra
    for bar in bars:
        height = bar.get_height()
        axes[i].text(
            bar.get_x() + bar.get_width()/2., 
            height + 0.01, # Un pequeño margen sobre la barra
            f'{height:.3f}', 
            ha='center', 
            va='bottom', 
            fontsize=10,
            fontweight='bold'
        )
    
    # --- AJUSTE CLAVE AQUÍ ---
    axes[i].set_ylim(0, 0.50) # Extendemos el eje Y hasta 0.5
    # -------------------------

    # Configuración de cada subplot
    axes[i].set_title(f'Distribución Binomial ($p = {p}$)', fontsize=14)
    axes[i].set_xlabel('Número de días con pérdida', fontsize=12)
    axes[i].set_xticks(x)
    axes[i].grid(axis='y', linestyle=':', alpha=0.6)

# Configuración general
axes[0].set_ylabel('Probabilidad $P(X=x_i)$', fontsize=12)
plt.tight_layout()


plt.savefig('distribucion_binomial_para_H.png', dpi=300, bbox_inches='tight')
# Mostrar gráfico
plt.show()
```
