```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
import scipy.stats as stats
import joblib
import math
import numpy as np
import itertools 
from itertools import product
from collections import Counter
import garpar as gp
import re
import skcriteria as skc
from skcriteria import pipelines
from skcriteria.preprocessing import invert_objectives, weighters, scalers
from skcriteria.madm import similarity, moora, electre
from skcriteria.ranksrev import RankInvariantChecker
```

```python
mer = joblib.load("./price_01-01-2010_to_01-10-2025.pklj")
mer= mer[3408:3808]#3408:3808
mer = mer.drop(columns=['BRIO', 'BRIO6','EDLH','EMDE','ESME','INAG','IRCP','PGR'])
mer = mer.ffill().apply(pd.to_numeric, errors='coerce').bfill()
mer = mer.infer_objects(copy=False)
pd.set_option('future.no_silent_downcasting', True)
mer
```

```python
base19 = gp.StocksSet.from_prices(
        mer[1:],
        weights=None,
        title="2025Merval19m",
        description=("Argentine stock market prices") )
base12 = gp.StocksSet.from_prices(
        mer[155:],
        weights=None,
        title="2025Merval12m",
        description=("Argentine stock market prices") )
base9 = gp.StocksSet.from_prices(
        mer[220:],
        weights=None,
        title="2025Merval9m",
        description=("Argentine stock market prices") )
base6 = gp.StocksSet.from_prices(
        mer[282:],
        weights=None,
        title="2025Merval6m",
        description=("Argentine stock market prices") )
base3 = gp.StocksSet.from_prices(
        mer[339:],
        weights=None,
        title="2025Merval3m",
        description=("Argentine stock market prices") )
```

```python
ms = gp.optimize.mean_variance.MVOptimizer( model='max_sharpe')
```

```python
# Diccionario de bases originales
bases = {
    "Sharpe 19m": base19,
    "Sharpe 12m": base12,
    "Sharpe 9m":  base9,
    "Sharpe 6m":  base6,
    "Sharpe 3m":  base3,
}

# Lista para guardar resultados
rows = []

for name, base in bases.items():

    # Optimización + pruning
    mopt = ms.optimize(base).weights_prune()


    # Guardar en lista de diccionarios
    rows.append({
        "Base": name,
        "stocks_number": len(mopt.stocks),
        "Cross_Entropy": mopt.div.cross_entropy(),
       #"KeyZang_Entropy": mopt.div.ke_zang_entropy(),
        "Zheng_Entropy": mopt.div.zheng_entropy(),
        "Div_Ratio": mopt.div.ratio(),
        "CV_MC":(mopt.div.mrc() *  mopt.weights).std() / (mopt.div.mrc() *  mopt.weights).mean(),
        "PDI": mopt.div.pdi(n_components=3, random_state=100)
    })

# Crear el DataFrame final
df_metrics = pd.DataFrame(rows).set_index("Base")

df_metrics
```

```python
mk = gp.optimize.mean_variance.Markowitz(target_return=0.001)
```

```python
# Diccionario de bases originales
basesm = {
    "Retorno mín. 19m": base19,
    "Retorno mín. 12m": base12,
    "Retorno mín. 9m":  base9,
    "Retorno mín. 6m":  base6,
    "Retorno mín. 3m":  base3,
}

# Lista para guardar resultados
rows = []

for name, base in basesm.items():
    # Optimización + pruning
    # Nota: asegúrate de que 'base' no tenga NaNs antes de optimizar si es posible
    mopt = mk.optimize(base).weights_prune()

    # Inicializamos el valor de PDI como nulo por si falla
    pdi_val = np.nan
    
    try:
        # Intentamos el cálculo del PDI
        pdi_val = mopt.div.pdi(n_components=3, random_state=100)
    except Exception as e:
        print(f"Advertencia: No se calculó PDI para {name}. Motivo: {e}")

    # Guardar en lista de diccionarios
    rows.append({
        "Base": name, 
        "stocks_number": len(mopt.stocks),
        "Cross_Entropy": mopt.div.cross_entropy(),
        "Zheng_Entropy": mopt.div.zheng_entropy(),
        "Div_Ratio": mopt.div.ratio(),
        "CV_MC": (mopt.div.mrc() * mopt.weights).std() / (mopt.div.mrc() * mopt.weights).mean(),
        "PDI": pdi_val
    })

# Crear el DataFrame final
df_metricsm = pd.DataFrame(rows).set_index("Base")

# Mostrar resultado
df_metricsm
```

```python
describe_metrics = metricssharpe.describe().transpose()
describe_metrics["kurt"]=metricssharpe.kurt() 
describe_metrics["skew"]=metricssharpe.skew() 
describe_metrics["Max/Min"]=describe_metrics["max"] / describe_metrics["min"]#Métrica maximo sobre el mínimo 
nombres_filas = {
    'stocks_number': 'Número de Acciones',
    'Cross_Entropy': 'Entropía Cruzada',
    'Zheng_Entropy': 'Entropía de Zheng',
    'Div_Ratio': 'Ratio de Diversificación',
    'CV_MC': 'CV MC',
    'PDI': 'Índice PDI'
}
# 2. Diccionario para las columnas (Estadísticos)
nombres_columnas = {
    'count': 'n',
    'mean': 'Media',
    'std': 'Desv. Est.',
    'min': 'Mín',
    'max': 'Máx',
    'kurt': 'Curtosis',
    'skew': 'Asimetría',
    'Max/Min': 'Máx/Mín'
}
describe_metrics.rename(index=nombres_filas).rename(columns=nombres_columnas).style.format(precision=3).to_latex(hrules=False)
```

```python
objectives = [
    max, #     'stocks_seleccionados', coincide con definición mayor número de stocks
    min, #     'Entropia_cruzada', mide el desorden al compararlo con una estrategia teórica de risk part equally
    max, #     'Entropia_Zheng', 
    max, #     'Ratio_diversific', mide la diversificación ganada por  mantener activos no correlacionados
    min, #     'CV_Contribuciones_marginales', si todos los activos tienen igual contribución marginal es un portfolio diversificado
    max, #     'Indice_diversific_port, si hay correlacióan alta el índice dará un valor bajo mientras que si no hay correlación el índice tenderá al valor N
]
dm = skc.mkdm(metricssharpe.to_numpy(), objectives, alternatives=metricssharpe.index, criteria=metricssharpe.columns)
```

```python
pipe_igual_vector = pipelines.mkpipe(
    invert_objectives.NegateMinimize(),
    scalers.VectorScaler(target="matrix"),
    weighters.EqualWeighter(),
    similarity.TOPSIS(),
)
pipe_igual_vector
result_igual_vector = pipe_igual_vector.evaluate(dm)
result_igual_vector
```

```python
result_igual_vector.extra_.similarity
```
